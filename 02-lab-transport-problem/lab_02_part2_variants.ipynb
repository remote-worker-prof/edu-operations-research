## 7. Вариант 6. Военный кейс: рационы и медкомплекты

Тип задачи: **закрытая**.

### Запасы

| Поставщик | Объём |
| --- | ---: |
| Узел A | 35 |
| Узел B | 25 |
| Узел C | 40 |

### Спрос

| Потребитель | Объём |
| --- | ---: |
| Точка 1 | 20 |
| Точка 2 | 15 |
| Точка 3 | 30 |
| Точка 4 | 35 |

### Матрица затрат

| Откуда / Куда | Точка 1 | Точка 2 | Точка 3 | Точка 4 |
| --- | ---: | ---: | ---: | ---: |
| Узел A | 4 | 6 | 60 | 7 |
| Узел B | 5 | 4 | 8 | 50 |
| Узел C | 6 | 5 | 4 | 6 |

Подсказка: маршруты `Узел A -> Точка 3` и `Узел B -> Точка 4` сделаны крайне дорогими. В отчёте отдельно поясните, удалось ли оптимальному плану их избежать.


```python
import numpy as np
from scipy.optimize import linprog

# Это шаблон для одного самостоятельного варианта.
# Сначала перенесите числа из условия, затем проверьте баланс,
# и только после этого запускайте решатель.

supply = np.array([np.nan, np.nan, np.nan], dtype=float)
demand = np.array([np.nan, np.nan, np.nan, np.nan], dtype=float)
costs = np.array([
    [np.nan, np.nan, np.nan, np.nan],
    [np.nan, np.nan, np.nan, np.nan],
    [np.nan, np.nan, np.nan, np.nan],
], dtype=float)

if np.isnan(supply).any() or np.isnan(demand).any() or np.isnan(costs).any():
    print("Заполните supply, demand и costs числами из выбранного варианта.")
    print("Если задача открытая, сначала сбалансируйте её фиктивным узлом.")
else:
    m, n = costs.shape
    c = costs.flatten()

    A_eq = []
    b_eq = []

    # Строки A_eq по поставщикам.
    for i in range(m):
        row = np.zeros(m * n)
        row[i * n : (i + 1) * n] = 1
        A_eq.append(row)
        b_eq.append(supply[i])

    # Строки A_eq по потребителям.
    for j in range(n):
        row = np.zeros(m * n)
        row[j::n] = 1
        A_eq.append(row)
        b_eq.append(demand[j])

    A_eq = np.array(A_eq, dtype=float)
    b_eq = np.array(b_eq, dtype=float)
    bounds = [(0, None)] * (m * n)

    result = linprog(c, A_eq=A_eq, b_eq=b_eq, bounds=bounds, method="highs")
    print("success:", result.success)
    print("message:", result.message)
    if result.success:
        # reshape() возвращает решение обратно к знакомой таблице перевозок.
        plan = result.x.reshape(m, n)
        plan[np.abs(plan) < 1e-9] = 0.0
        print("plan =")
        print(plan)
        print("min_cost =", result.fun)
        print("row sums =", plan.sum(axis=1))
        print("column sums =", plan.sum(axis=0))
```
